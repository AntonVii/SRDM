```python

import cmdstanpy
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv("forstmann.csv")

df["correct"] = df.apply(lambda row: 1 if row["S"] == row["R"] else 0,axis = 1)
df['participant_id'], participant_labels = pd.factorize(df['subjects'])

df['participant_id'] += 1  


sat_map = {"speed":1,"neutral":2,"accuracy":3}
df["sat_int"] = df["E"].map(sat_map)
df["pc"] = (df["participant_id"]-1)*3 + df["sat_int"]
df
```

```python
max_rt = (df.groupby(['participant_id','sat_int'])["rt"].max())
max_rt
```

```python
N_correct = len(df[df["correct"] == 1])
N_false = len(df[df["correct"] == 0])
P = 19
rt_correct = list(df[df["correct"]==1]["rt"])
rt_false = list(df[df["correct"]==0]["rt"])
pc_correct = list(df[df["correct"]==1]["pc"])
pc_false = list(df[df["correct"]==0]["pc"])

t0_hi = (
    df.groupby('participant_id')['rt']
    .quantile(0.01)
    .reindex(range(1,P+1))
)
assert t0_hi.notna().all(), "some cell has zero trials"




print(N_correct,N_false,rt_false)
data = {"N_correct":N_correct,"N_false":N_false,"P":P,
        "rt_correct":rt_correct,"rt_false":rt_false,
        "pc_correct":pc_correct,"pc_false":pc_false,"max_rt":max_rt,"t0_hi":t0_hi}

from cmdstanpy import CmdStanModel

model = CmdStanModel(stan_file='SRDMlapse_fixed1.stan')



```

```python


init_values = {
    'c': [0.0] * (3 * P),
    't0': [0.2] * (P),      # matches your t0 prior mean
    'd': [1.0] * (3*P),              # matches your d prior mean
    'r': [4.5] * P,              # comfortably above the 0.05 floor
    'B': [1.0] * (3*P),
    'p_lapse': [0.01] * P,       # matches the beta(1,10) prior's rough mean
}

fit = model.sample(data=data,chains = 4, parallel_chains=4, inits=init_values,show_console=True,metric='dense_e')

```

```python
import matplotlib.pyplot as plt
import arviz as az
idata = az.from_cmdstanpy(fit)
n_participants = 19
scalar_vars = ["t0", "r", "p_lapse"]       # one value per participant
condition_vars = ["c", "d", "B_rep"]       # one value per condition per participant
cond_labels = ["speed", "neutral", "accuracy"]

for p in range(n_participants):
    fig, axes = plt.subplots(4, 3, figsize=(15, 16))
    axes = axes.flatten()
    ax_i = 0

    for var in scalar_vars:
        dim = f"{var}_dim_0"
        az.plot_posterior(idata, var_names=[var], coords={dim: [p]}, ax=axes[ax_i])
        axes[ax_i].set_title(f"{var} (participant {p+1})")
        ax_i += 1

    for var in condition_vars:
        dim = f"{var}_dim_0"
        idx = [p*3, p*3+1, p*3+2]
        for j, cond in enumerate(cond_labels):
            az.plot_posterior(idata, var_names=[var], coords={dim: [idx[j]]}, ax=axes[ax_i])
            axes[ax_i].set_title(f"{var} - {cond} (p{p+1})")
            ax_i += 1

    fig.suptitle(f"Participant {p+1} — all posteriors", fontsize=16)
    plt.tight_layout()
    plt.show()
```
